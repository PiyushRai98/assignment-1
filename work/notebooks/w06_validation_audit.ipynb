## 2. My model under an honest split (before/after)

*Re-run your Week-5 model under a grouped or time-aware split. Show both numbers.*

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression

# --- MOCK DATA FOR DEMONSTRATION ---
# Replace this block with your actual data loading
np.random.seed(42)
dates = pd.date_range('2023-01-01', periods=100)
df = pd.DataFrame({
    'date_column': dates,
    'feature_1': np.random.randn(100),
    'feature_2': np.random.randn(100),
    'target': np.random.randn(100)
})
df = df.sort_values('date_column')
X = df[['feature_1', 'feature_2']]
y = df['target']
# -----------------------------------

# 1. Random Split (Week 5 Style Mockup)
# In practice, use your specific Week 5 model and split here
random_rmse = 0.85 # Example placeholder

# 2. Time-Aware Split
tscv = TimeSeriesSplit(n_splits=3)
rmse_scores = []

for train_index, test_index in tscv.split(X):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    model = LinearRegression()
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    rmse_scores.append(np.sqrt(mean_squared_error(y_test, preds)))

time_aware_rmse = np.mean(rmse_scores)

print(f"Random Split RMSE: {random_rmse}")
print(f"Time-Aware Split RMSE: {time_aware_rmse:.4f}")
```

```python
# Leakage Audit: Correlation check between features and target
# and checking if any feature depends on future information.

# Ensure we only correlate numeric columns
numeric_df = df.select_dtypes(include=[np.number])
correlations = numeric_df.corr()['target'].sort_values(ascending=False)

print("Top Correlations (Check for >0.95 or suspicious features):")
display(correlations.head(10))

# Check for target leakage variables based on naming conventions
suspicious_cols = [col for col in df.columns if 'target' in col.lower() and col != 'target']
if suspicious_cols:
    print(f"Suspicious columns found: {suspicious_cols}")
else:
    print("No obviously named target leakage columns found.")
```
