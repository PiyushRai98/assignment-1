## 4. Monitoring / retrain triggers

* **Drift Trigger**: If the average prediction score drops by >15% over a week.
* **Quality Trigger**: If the click-through rate (CTR) falls below the 2% baseline.

```python
def check_triggers(current_ctr, baseline_ctr=0.02):
    if current_ctr < baseline_ctr:
        print(f"ALERT: Quality Trigger! CTR at {current_ctr:.2%}")
    else:
        print(f"Monitoring: CTR at {current_ctr:.2%} is within bounds.")

check_triggers(0.018)  # Example trigger
```
